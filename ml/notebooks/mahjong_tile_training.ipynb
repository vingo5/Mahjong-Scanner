```python
from roboflow import Roboflow
from google.colab import userdata
api_key = userdata.get('ROBOFLOW_API_KEY')

rf = Roboflow(api_key)
project = rf.workspace("project-xv49e").project("mahjong-x5dzz")
dataset = project.version(2).download("yolov8")
```

```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

results = model.train(
    data=f"{dataset.location}/data.yaml",
    epochs=100,
    imgsz=640,
    batch=16,
    patience=20,
    name="mahjong_tiles_v1",
    project="/content/drive/MyDrive/mahjong_training_runs",  # writes directly to Drive
    save_period=5,        # force a checkpoint every 5 epochs, not just at the end
    device=0,
)
```

```python
metrics = model.val()
print(f"mAP50-95: {metrics.box.map:.4f}")
print(f"mAP50: {metrics.box.map50:.4f}")
```
